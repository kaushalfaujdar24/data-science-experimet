```python
#40
matrix = np.arange(1, 17).reshape(4, 4)
transpose_mat = matrix.T
row_sums = np.sum(matrix, axis=1)
col_sums = np.sum(matrix, axis=0)
overall_mean = np.mean(matrix)
overall_max = np.max(matrix)
overall_min = np.min(matrix)
central_matrix = matrix[1:3, 1:3]
print("Original Matrix", matrix)
print("Transpose", transpose_mat)
print("Row-wise Sums", row_sums)
print("Column-wise Sums", col_sums)
print("Mean", overall_mean)
print("Maximum Value", overall_max)
print("Minimum Value", overall_min)
print("Central Matrix", central_matrix)
```

```python
#41
data = np.random.randn(20)
mean_val = np.mean(data)
std_val = np.std(data)
min_val = np.min(data)
max_val = np.max(data)
sorted_data = np.sort(data)
cleaned_data = np.where(data < 0, 0, data)
print("data: ",np.round(data,2))
print("mean: ",np.round(mean_val,2))
print("std: ",np.round(std_val,2))
print("min: ",np.round(min_val,2))
print("max: ",np.round(max_val,2))
print("sorted: ",np.round(sorted_data,2))
print("cleaned: ",np.round(cleaned_data,2))
```

```python
#42
A = np.array([[2, 4, 6],
              [1, 3, 5],
              [7, 8, 9]])
B = np.array([[9, 8, 7],
              [6, 5, 4],
              [3, 2, 1]])
elem_add = A + B
elem_mult = A * B
mat_mult = np.dot(A, B)
A_transpose = A.T
mat_mult_max = np.max(mat_mult)
mat_mult_min = np.min(mat_mult)
print("Matrix A:", A)
print("Matrix B:", B)
print("Element-wise Addition:", elem_add)
print("Element-wise Multiplication:", elem_mult)
print("Matrix Multiplication:", mat_mult)
print("Transpose of Matrix A:", A_transpose)
print("Max of Resulting Product Matrix:", mat_mult_max)
print("Min of Resulting Product Matrix:", mat_mult_min)
```
